````python
from huggingface_hub import InferenceClient
from pydantic import BaseModel
import json, os

class Fact(BaseModel):
    entity: str
    attribute: str
    value: str
    chapter: int
    day: str | None = None
    quote: str

class Extraction(BaseModel):
    facts: list[Fact]

client = InferenceClient(token=os.environ["HF_TOKEN"])

def extract(text, chapter, model="Qwen/Qwen2.5-72B-Instruct"):
    resp = client.chat_completion(
        model=model,
        messages=[
            {"role": "system", "content": "Extract story facts. Return ONLY JSON matching: " + json.dumps(Extraction.model_json_schema())},
            {"role": "user", "content": f"Chapter {chapter}:\n{text}"},
        ],
        max_tokens=1500, temperature=0,
    )
    raw = resp.choices[0].message.content.strip().removeprefix("```json").removesuffix("```").strip()
    return Extraction.model_validate_json(raw)

ch2 = "On Monday morning Aarav boarded the train in Delhi, clutching his mother's old watch."
result = extract(ch2, 2)
for f in result.facts:
    print(f)
````

````python
import time

_orig_generate = gclient.models.generate_content

def _retrying_generate(*args, **kwargs):
    for attempt in range(6):
        try:
            return _orig_generate(*args, **kwargs)
        except Exception as e:
            busy = any(code in str(e) for code in ("503", "429", "UNAVAILABLE"))
            if busy and attempt < 5:
                wait = 5 * (attempt + 1)
                print(f"Server busy, retrying in {wait}s...")
                time.sleep(wait)
            else:
                raise

gclient.models.generate_content = _retrying_generate
````

````python
def check(bible, text, chapter):
    prompt = ("You are a strict continuity editor. Below is the established story bible (facts from EARLIER chapters) "
              "and a NEW chapter. Report ONLY genuine contradictions where the new chapter conflicts with a fact in the bible: "
              "impossible travel times, changed physical traits, broken world rules, or a character knowing something "
              "the bible says they were never told. Rules: old_quote must come from the bible and new_quote from the new chapter, "
              "and they must be different passages. Do NOT report a statement that is consistent with the bible, "
              "restates it, or merely says a character does NOT know something. When unsure, do not report. "
              "Return ONLY JSON matching: " + json.dumps(Conflicts.model_json_schema())
              + "\nIf there are none, return {\"conflicts\": []}.\n\n"
              + "BIBLE:\n" + json.dumps([f.model_dump() for f in bible])
              + f"\n\nNEW CHAPTER {chapter}:\n{text}")
    r = gclient.models.generate_content(model=MODEL, contents=prompt,
        config={"response_mime_type": "application/json", "temperature": 0})
    found = Conflicts.model_validate_json(r.text).conflicts
    return [c for c in found if c.old_quote.strip() != c.new_quote.strip()]

bible = []
for i, ch in enumerate(chapters, 1):
    if bible:
        for c in check(bible, ch, i):
            print(f"⚠ Ch{i} [{c.type}] {c.explanation}\n   was: {c.old_quote!r}\n   now: {c.new_quote!r}\n")
    bible.extend(extract(ch, i).facts)
print(f"Bible has {len(bible)} facts")
````

````python
import gradio as gr

state = {"bible": [], "chapter": 0}

def load_demo():
    """Pre-load chapters 1-4 so the demo starts with a populated bible."""
    state["bible"], state["chapter"] = [], 0
    for i, ch in enumerate(chapters[:3], 1):          # chapters 1-3 are the "established" story
        state["bible"].extend(extract(ch, i).facts)
        state["chapter"] = i
    return chapters[3], table(), f"Loaded chapters 1-3 ({len(state['bible'])} facts). Chapter 4 is in the editor: click Check."

def table():
    return [[f.entity, f.attribute, f.value, f.chapter, f.day or ""] for f in state["bible"]]

def run_check(text):
    if not text.strip():
        return "Paste a chapter first.", table()
    state["chapter"] += 1
    n = state["chapter"]
    flags = check(state["bible"], text, n) if state["bible"] else []
    state["bible"].extend(extract(text, n).facts)
    if not flags:
        report = f"✅ Chapter {n}: no contradictions found."
    else:
        report = f"### ⚠ Chapter {n}: {len(flags)} contradiction(s)\n\n" + "\n\n".join(
            f"**{c.type}**: {c.explanation}\n\n> was: *{c.old_quote}*\n>\n> now: *{c.new_quote}*" for c in flags)
    return report, table()

with gr.Blocks(title="Story Bible") as app:
    gr.Markdown("# 📖 Story Bible\nWrite a chapter, and the AI checks it against everything established so far.")
    with gr.Row():
        with gr.Column():
            editor = gr.Textbox(label="Chapter text", lines=14, placeholder="Paste or write a chapter...")
            with gr.Row():
                check_btn = gr.Button("Check chapter", variant="primary")
                demo_btn = gr.Button("Load demo story")
        with gr.Column():
            report = gr.Markdown("Flags will appear here.")
            bible_view = gr.Dataframe(headers=["Entity", "Attribute", "Value", "Chapter", "Day"],
                                      label="Live story bible", interactive=False)
    check_btn.click(run_check, editor, [report, bible_view])
    demo_btn.click(load_demo, None, [editor, bible_view, report])

app.launch(inline=False)
````

````python
from concurrent.futures import ThreadPoolExecutor

def load_demo():
    with open("Story_Wave_demo.json", encoding="utf-8") as fp:
        data = json.load(fp)
    state["bible"] = [Fact(**d) for d in data if d["chapter"] <= 3]
    state["chapter"] = 3
    return chapters[3], table(), f"Loaded chapters 1-3 ({len(state['bible'])} facts). Chapter 4 is in the editor: click Check."

def run_check(text):
    if not text.strip():
        return "Paste a chapter first.", table()
    state["chapter"] += 1
    n = state["chapter"]
    with ThreadPoolExecutor(max_workers=2) as ex:
        check_job = ex.submit(check, state["bible"], text, n) if state["bible"] else None
        extract_job = ex.submit(extract, text, n)
        flags = check_job.result() if check_job else []
        new_facts = extract_job.result().facts
    state["bible"].extend(new_facts)
    if not flags:
        report = f"✅ Chapter {n}: no contradictions found."
    else:
        report = f"### ⚠ Chapter {n}: {len(flags)} contradiction(s)\n\n" + "\n\n".join(
            f"**{c.type}**: {c.explanation}\n\n> was: *{c.old_quote}*\n>\n> now: *{c.new_quote}*" for c in flags)
    return report, table()
````

````python
import mimetypes, tempfile, wave
from google.genai import types

TTS_MODEL = "PASTE-TTS-MODEL-NAME-HERE"
last_voice = {"emotion": "calm", "intensity": 5}


class VoiceReport(BaseModel):
    transcript: str
    primary_emotion: str
    secondary_emotion: str
    intensity: int  # 1-10
    valence: Literal["negative", "neutral", "positive"]
    energy: Literal["low", "medium", "high"]
    delivery_cues: str  # pace, pitch, pauses, tremor, breathiness
    words_vs_tone_mismatch: bool
    mismatch_note: str
    summary: str


def analyze_voice(path: str) -> VoiceReport:
    mime = mimetypes.guess_type(path)[0] or "audio/wav"
    with open(path, "rb") as fp:
        audio = types.Part.from_bytes(data=fp.read(), mime_type=mime)
    prompt = (
        "Listen to this recording of a person speaking. Judge emotion from HOW it is said (pace, pitch, pauses, "
        "tremor, breathiness, volume), not just the words. Give a transcript, primary and secondary emotion, "
        "intensity 1-10, valence, energy, delivery cues, and whether the tone contradicts the words (e.g. 'I'm fine' "
        "said in a shaky voice), with a note. These are interpretations, so avoid overclaiming. "
        "Return ONLY JSON matching: " + json.dumps(VoiceReport.model_json_schema())
    )
    return VoiceReport.model_validate_json(call_llm([audio, prompt]))


def run_voice(path):
    if not path:
        return "Record or upload a clip first (under a minute works best)."
    r = analyze_voice(path)
    last_voice.update(emotion=r.primary_emotion, intensity=r.intensity)
    bar = "█" * r.intensity + "░" * (10 - r.intensity)
    mismatch = f"⚠️ **Tone vs words mismatch:** {r.mismatch_note}\n\n" if r.words_vs_tone_mismatch else ""
    return (f"**Heard:** *{r.transcript}*\n\n**{r.primary_emotion}** (also {r.secondary_emotion})  `{bar}` {r.intensity}/10\n\n"
            f"Valence: {r.valence} · Energy: {r.energy}\n\nDelivery: {r.delivery_cues}\n\n{mismatch}{r.summary}")


def speak(text, emotion, intensity):
    if not text.strip():
        return None
    style = (f"Say this in a natural human way with {emotion}, intensity {int(intensity)} out of 10: "
             "real pacing, pauses and breath, not theatrical. Text: ")
    for attempt in range(3):
        try:
            r = client.models.generate_content(
                model=TTS_MODEL, contents=style + text,
                config=types.GenerateContentConfig(
                    response_modalities=["AUDIO"],
                    speech_config=types.SpeechConfig(voice_config=types.VoiceConfig(
                        prebuilt_voice_config=types.PrebuiltVoiceConfig(voice_name="Kore")))))
            pcm = r.candidates[0].content.parts[0].inline_data.data
            break
        except Exception as e:  # noqa: BLE001
            print("TTS failed:", str(e)[:80]); time.sleep(5)
    else:
        raise RuntimeError("Speech model busy or unavailable.")
    path = os.path.join(tempfile.gettempdir(), f"ai_voice_{int(time.time())}.wav")
    with wave.open(path, "wb") as w:
        w.setnchannels(1); w.setsampwidth(2); w.setframerate(24000); w.writeframes(pcm)
    return path


def speak_matching(text):
    return speak(text, last_voice["emotion"], last_voice["intensity"])
````

````python
import edge_tts, asyncio
from concurrent.futures import ThreadPoolExecutor

def prosody(emotion, intensity):
    e, k = emotion.lower(), intensity / 10
    if any(w in e for w in ("grief", "sad", "despair", "sorrow", "helpless", "hurt", "lonely", "melanchol")):
        rate, pitch, vol = -30*k, -15*k, -20*k
    elif any(w in e for w in ("anger", "angry", "rage", "fury", "frustrat")):
        rate, pitch, vol = 10*k, 5*k, 25*k
    elif any(w in e for w in ("fear", "anxi", "panic", "nervous", "worr")):
        rate, pitch, vol = 20*k, 15*k, 0
    elif any(w in e for w in ("joy", "happy", "excite", "cheer", "delight")):
        rate, pitch, vol = 15*k, 10*k, 15*k
    else:
        rate, pitch, vol = 0, 0, 0
    return f"{rate:+.0f}%", f"{pitch:+.0f}Hz", f"{vol:+.0f}%"

def speak(text, emotion, intensity):
    if not text.strip():
        return None
    rate, pitch, vol = prosody(emotion, int(intensity))
    path = os.path.join(tempfile.gettempdir(), f"ai_voice_{int(time.time())}.mp3")
    coro = edge_tts.Communicate(text, "en-IN-NeerjaNeural", rate=rate, pitch=pitch, volume=vol).save(path)
    with ThreadPoolExecutor(1) as ex:
        ex.submit(asyncio.run, coro).result()
    return path

from IPython.display import Audio
Audio(filename=speak("I'm fine. Honestly, it's fine.", "suppressed grief", 8))
````

````python
class Branch(BaseModel):
    title: str
    scene: str

class Branches(BaseModel):
    branches: list[Branch]

gen_state = {"results": [], "labels": [], "chapter": 0}


def generate_branches(bible, direction, emotion, intensity, sabotage):
    extra = ("\nFor TESTING ONLY: make Branch 3 quietly break exactly one established fact from the bible "
             "(for example a changed trait or a broken world rule), without drawing attention to it." if sabotage else "")
    prompt = (
        "You are a co-writer. Using the story bible, write THREE different branching versions of the next scene "
        "(110-160 words each). Each branch takes the story in a clearly different direction. Respect every fact in "
        "the bible and keep each character's voice true to who they are (age, background, how they spoke before). "
        "Include dialogue in double quotes, with who is speaking. "
        f"What happens next: {direction}\n"
        f"Target emotion: {emotion}, intensity {intensity}/10. The underlying feeling may differ from what characters "
        "say aloud. Give each branch a short title." + extra
        + " Return ONLY JSON matching: " + json.dumps(Branches.model_json_schema())
        + "\n\nBIBLE:\n" + json.dumps([f.model_dump() for f in bible])
    )
    return Branches.model_validate_json(call_llm(prompt)).branches[:3]


def evaluate_branch(branch, bible, n):
    try:
        flags = check(bible, branch.scene, n)
    except Exception as e:
        print("check failed:", str(e)[:80]); flags = None
    try:
        emo = analyze_emotion(bible, branch.scene, n)
    except Exception as e:
        print("emotion failed:", str(e)[:80]); emo = None
    return {"branch": branch, "flags": flags, "emo": emo}


def run_generate(direction, emotion, intensity, sabotage):
    if not state["bible"]:
        return "Load the demo story first (Consistency tab) so there is a bible to write from.", gr.update()
    n = state["chapter"] + 1
    branches = generate_branches(state["bible"], direction, emotion, int(intensity), sabotage)
    with ThreadPoolExecutor(max_workers=3) as ex:
        results = list(ex.map(lambda b: evaluate_branch(b, state["bible"], n), branches))
    labels = [f"{chr(65 + i)}: {r['branch'].title}" for i, r in enumerate(results)]
    gen_state.update(results=results, labels=labels, chapter=n)

    parts = []
    for label, r in zip(labels, results):
        scene = "\n".join("> " + ln for ln in r["branch"].scene.splitlines())
        if r["flags"] is None:
            cont = "❔ Continuity check failed, try again."
        elif not r["flags"]:
            cont = "✅ No contradictions with the story bible."
        else:
            cont = f"⚠️ **{len(r['flags'])} contradiction(s):**\n" + "\n".join(
                f"- {c.explanation}" for c in r["flags"])
        if r["emo"] and r["emo"].lines:
            peak = max(l.intensity for l in r["emo"].lines)
            drifts = sum(l.voice_match == "drifts" for l in r["emo"].lines)
            emo = f"Peak emotion {peak}/10 · voice drifts: {drifts}\n\n*{r['emo'].arc_summary}*"
        else:
            emo = "No scored dialogue in this branch."
        parts.append(f"### Branch {label}\n\n{scene}\n\n**Continuity:** {cont}\n\n**Emotion:** {emo}")
    return "\n\n---\n\n".join(parts), gr.update(choices=labels, value=None)


def _find(choice):
    if choice in gen_state["labels"]:
        return gen_state["results"][gen_state["labels"].index(choice)]
    return None


def commit_branch(choice):
    r = _find(choice)
    if not r:
        return "Pick a branch first.", table()
    n = gen_state["chapter"]
    state["bible"].extend(extract(r["branch"].scene, n).facts)
    state["chapter"] = n
    warn = f" It was committed with {len(r['flags'])} contradiction(s) still in it." if r["flags"] else ""
    return f"✅ Committed to chapter {n}. The bible now has {len(state['bible'])} facts.{warn}", table()


def speak_branch(choice):
    r = _find(choice)
    if not r:
        return None, "Pick a branch first."
    if not (r["emo"] and r["emo"].lines):
        return None, "That branch has no dialogue to speak."
    top = max(r["emo"].lines, key=lambda l: l.intensity)
    path = speak(top.line, top.underlying_emotion, top.intensity)
    return path, f"🔊 Speaking {top.speaker}'s line with the **underlying** emotion: {top.underlying_emotion} ({top.intensity}/10)."
````

````python
@dataclass
class StoryBible:
    facts: list[Fact] = field(default_factory=list)   # never a shared/global default

    def add(self, new_facts: list[Fact]) -> None:
        self.facts.extend(new_facts)

    def reset(self) -> None:
        self.facts.clear()

    def to_df(self) -> pd.DataFrame:
        cols = ["entity", "attribute", "value", "chapter", "day"]
        if not self.facts:
            return pd.DataFrame(columns=cols)
        return pd.DataFrame([asdict(f) for f in self.facts])[cols]

# Universe store starts EMPTY. Nothing from the demo is loaded here.
BIBLES: dict[str, StoryBible] = {}

def get_bible(universe: str) -> StoryBible:
    return BIBLES.setdefault(universe, StoryBible())

def fork_universe(src: str, new_name: str) -> str:
    BIBLES[new_name] = deepcopy(get_bible(src))   # deepcopy so universes don't share facts
    return new_name
````

````python
import os, json
from pydantic import BaseModel
from google import genai
from google.genai import types

MODEL = "gemini-2.5-flash"   # use whichever model your Story_Wave_model.py already uses
client = genai.Client(api_key=os.environ["GEMINI_API_KEY"])

class FactOut(BaseModel):
    entity: str
    attribute: str          # location, age, appearance, occupation, relationship, knowledge, rule
    value: str
    day: str | None = None
    quote: str

class ExtractionOut(BaseModel):
    facts: list[FactOut]

EXTRACT_PROMPT = """Extract story facts ONLY from the chapter text below.
Do not use outside knowledge or any other story.
Each fact needs: entity (character/place/object), attribute, value, day (if stated), and a short verbatim quote from the text.
Skip generic narration with no concrete fact."""

def extract_facts(text: str, chapter: int) -> list[Fact]:
    resp = client.models.generate_content(
        model=MODEL,
        contents=f"{EXTRACT_PROMPT}\n\nChapter {chapter}:\n{text}",
        config=types.GenerateContentConfig(
            response_mime_type="application/json",
            response_schema=ExtractionOut,
            temperature=0,
        ),
    )
    try:
        data = ExtractionOut.model_validate_json(resp.text)
    except Exception:
        # one retry
        resp = client.models.generate_content(
            model=MODEL,
            contents=f"{EXTRACT_PROMPT}\n\nReturn valid JSON only.\n\nChapter {chapter}:\n{text}",
            config=types.GenerateContentConfig(
                response_mime_type="application/json",
                response_schema=ExtractionOut,
                temperature=0,
            ),
        )
        data = ExtractionOut.model_validate_json(resp.text)
    return [Fact(f.entity, f.attribute, f.value, chapter, f.day, f.quote) for f in data.facts]
````

````python
import os, json, getpass
from dataclasses import dataclass, field, asdict
from copy import deepcopy
import pandas as pd
from pydantic import BaseModel
from google import genai
from google.genai import types

if not os.environ.get("GEMINI_API_KEY"):
    os.environ["GEMINI_API_KEY"] = getpass.getpass("Gemini API key: ")

MODEL = "gemini-2.5-flash"   # match the model in your Story_Wave_model.py
client = genai.Client(api_key=os.environ["GEMINI_API_KEY"])

# ---------- data ----------
@dataclass
class Fact:
    entity: str
    attribute: str
    value: str
    chapter: int
    day: str | None = None
    quote: str = ""

@dataclass
class StoryBible:
    facts: list[Fact] = field(default_factory=list)
    def add(self, new): self.facts.extend(new)
    def reset(self): self.facts.clear()
    def to_df(self):
        cols = ["entity", "attribute", "value", "chapter", "day"]
        if not self.facts:
            return pd.DataFrame(columns=cols)
        return pd.DataFrame([asdict(f) for f in self.facts])[cols]

BIBLES: dict[str, StoryBible] = {}
def get_bible(u): return BIBLES.setdefault(u, StoryBible())
def fork_universe(src, new):
    BIBLES[new] = deepcopy(get_bible(src)); return new

# ---------- schemas ----------
class FactOut(BaseModel):
    entity: str
    attribute: str
    value: str
    day: str | None = None
    quote: str
class ExtractionOut(BaseModel):
    facts: list[FactOut]
class FlagOut(BaseModel):
    entity: str
    explanation: str
    old_quote: str
    new_quote: str
    old_chapter: int
    new_chapter: int
class FlagsOut(BaseModel):
    flags: list[FlagOut]

def _call(prompt, schema):
    cfg = types.GenerateContentConfig(
        response_mime_type="application/json",
        response_schema=schema, temperature=0)
    for attempt in range(2):                      # one retry
        try:
            r = client.models.generate_content(model=MODEL, contents=prompt, config=cfg)
            return schema.model_validate_json(r.text)
        except Exception as e:
            if attempt == 1: raise

EXTRACT_PROMPT = """Extract story facts ONLY from the chapter text below.
Do not use outside knowledge or any other story.
Each fact: entity (character/place/object), attribute (location, age, appearance,
occupation, relationship, knowledge, rule), value, day (if stated), and a short
verbatim quote. Skip generic narration with no concrete fact."""

CHECK_PROMPT = """You are a continuity editor. Compare NEW facts against EXISTING facts.
Report only true contradictions (same entity, incompatible values at the same time).
Do not flag facts that can both be true or that change legitimately over time.
Return an empty list if there are none."""

def extract_facts(text: str, chapter: int) -> list[Fact]:
    out = _call(f"{EXTRACT_PROMPT}\n\nChapter {chapter}:\n{text}", ExtractionOut)
    return [Fact(f.entity, f.attribute, f.value, chapter, f.day, f.quote) for f in out.facts]

def find_conflicts(bible: StoryBible, new_facts: list[Fact]) -> list[str]:
    if not bible.facts or not new_facts:
        return []
    names = {f.entity.lower() for f in new_facts}
    related = [f for f in bible.facts if f.entity.lower() in names]
    if not related:
        return []
    payload = {"existing": [asdict(f) for f in related],
               "new": [asdict(f) for f in new_facts]}
    out = _call(f"{CHECK_PROMPT}\n\n{json.dumps(payload, ensure_ascii=False)}", FlagsOut)
    return [f"⚠️ {f.entity}: {f.explanation} "
            f"(ch {f.old_chapter}: \"{f.old_quote}\" vs ch {f.new_chapter}: \"{f.new_quote}\")"
            for f in out.flags]

def check_chapter(universe: str, text: str, chapter: int):
    text = (text or "").strip()
    if not text:
        return []
    bible = get_bible(universe)
    new_facts = extract_facts(text, chapter)
    flags = find_conflicts(bible, new_facts)
    bible.add(new_facts)
    return flags

# ---------- guard against stubs ----------
import inspect
assert "Gemini" not in inspect.getsource(extract_facts) and "_call" in inspect.getsource(extract_facts), "stub still active"
assert "text[:20]" not in inspect.getsource(extract_facts), "stub still active"
print("Real functions loaded.")
````
